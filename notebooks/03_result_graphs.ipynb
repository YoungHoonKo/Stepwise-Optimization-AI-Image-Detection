```python
import pandas as pd
import matplotlib.pyplot as plt

# 데이터
df = pd.DataFrame({
    "model": [
        "Phase1\n(CIFAKE)",
        "Phase3\n(AI_REAL)",
        "Phase4\n(AI_REAL)",
        "Phase5\n(AI_REAL)"
    ],
    "accuracy": [
        0.188389,
        0.500000,
        0.440778,
        0.875556
    ],
    "loss": [
        11.893752,
        1.169009,
        4.213066,
        0.332260
    ]
})

# Figure 생성 (가로 2개 차트)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ---------------- Accuracy ----------------
bars1 = axes[0].bar(df["model"], df["accuracy"])

axes[0].set_title("ResNet18 Accuracy Comparison")
axes[0].set_ylabel("Accuracy")
axes[0].set_ylim(0, 1.0)
axes[0].grid(axis="y", linestyle="--", alpha=0.5)

for bar in bars1:
    h = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width()/2,
        h + 0.02,
        f"{h:.3f}",
        ha="center"
    )

# ---------------- Loss ----------------
bars2 = axes[1].bar(df["model"], df["loss"])

axes[1].set_title("ResNet18 Loss Comparison")
axes[1].set_ylabel("Loss")
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

for bar in bars2:
    h = bar.get_height()
    axes[1].text(
        bar.get_x() + bar.get_width()/2,
        h + 0.2,
        f"{h:.3f}",
        ha="center"
    )

plt.suptitle("Phase-wise Performance of ResNet18", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ===== 결과 입력 =====
individual_models = [
    "Phase1\n(CIFAKE)",
    "Phase3\n(AI vs Real)",
    "Phase4\n(AI vs Human)",
    "Phase5\n(Faces)"
]

individual_acc = [
    0.188389,
    0.500000,
    0.440778,
    0.875556
]

cumulative_acc = 0.883200

# ===== 그래프 =====
plt.figure(figsize=(12, 6))

# 개별 모델 (파란색)
bars1 = plt.bar(
    individual_models,
    individual_acc,
    color="tab:blue",
    alpha=0.8,
    label="Individual Dataset Models"
)

# 누적 모델 (주황색)
bars2 = plt.bar(
    ["Cumulative\nTraining Model"],
    [cumulative_acc],
    color="tab:orange",
    alpha=0.9,
    label="Cumulative Training Model"
)

# 값 표시
for bar in list(bars1) + list(bars2):
    h = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        h + 0.015,
        f"{h:.3f}",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

# 기준선 (50%)
plt.axhline(
    y=0.5,
    color="gray",
    linestyle="--",
    alpha=0.6
)

# 그래프 설정
plt.title(
    "Comparison of Test Accuracy by Model (SynthBuster Dataset)",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel("Test Accuracy", fontsize=14)
plt.ylim(0, 1.1)

plt.grid(axis="y", alpha=0.25)

plt.legend(
    loc="lower right",
    fontsize=11
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# =========================
# 누적 학습 모델 비교
# =========================
models = [
    "Cumulative\nCNN",
    "Cumulative\nResNet18"
]

accuracy = [
    0.785,
    0.883
]

colors = [
    "tab:blue",
    "tab:orange"
]

# =========================
# 그래프 생성
# =========================
plt.figure(figsize=(8, 6))

bars = plt.bar(
    models,
    accuracy,
    color=colors,
    alpha=0.85
)

# 값 표시
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.01,
        f"{height:.3f}",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold"
    )

# 기준선
plt.axhline(
    y=0.5,
    color="gray",
    linestyle="--",
    linewidth=2,
    alpha=0.6
)

# 범례
legend_elements = [
    Patch(facecolor="tab:blue",
          label="Cumulative CNN"),
    Patch(facecolor="tab:orange",
          label="Cumulative ResNet18")
]

plt.legend(
    handles=legend_elements,
    loc="lower right",
    fontsize=11
)

# 제목
plt.title(
    "Cumulative CNN vs Cumulative ResNet18",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Test Accuracy",
    fontsize=15
)

plt.ylim(0, 1.1)

plt.grid(
    axis="y",
    alpha=0.25
)

improvement = (0.883 - 0.785) * 100
print(f"Accuracy Improvement: {improvement:.1f}%p")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# =========================
# 누적 모델 결과
# =========================
cnn_acc = 0.785
resnet_acc = 0.883

models = [
    "Cumulative\nCNN",
    "Cumulative\nResNet18"
]

accuracy = [
    cnn_acc,
    resnet_acc
]

colors = [
    "tab:blue",
    "tab:orange"
]

improvement = (resnet_acc - cnn_acc) * 100  # %p

# =========================
# 그래프
# =========================
plt.figure(figsize=(8, 6))

bars = plt.bar(
    models,
    accuracy,
    color=colors,
    alpha=0.85,
    width=0.6
)

# 값 표시
for bar in bars:
    h = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        h + 0.01,
        f"{h:.3f}",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

# 차이 표시 화살표
x1 = bars[0].get_x() + bars[0].get_width()/2
x2 = bars[1].get_x() + bars[1].get_width()/2

y_arrow = 0.95

plt.annotate(
    "",
    xy=(x2, y_arrow),
    xytext=(x1, y_arrow),
    arrowprops=dict(
        arrowstyle="<->",
        lw=2
    )
)

plt.text(
    (x1 + x2) / 2,
    y_arrow + 0.02,
    f"+{improvement:.1f}%p",
    ha="center",
    fontsize=14,
    fontweight="bold"
)

# 기준선
plt.axhline(
    y=0.5,
    color="gray",
    linestyle="--",
    alpha=0.6
)

# 범례
legend_elements = [
    Patch(facecolor="tab:blue", label="Cumulative CNN"),
    Patch(facecolor="tab:orange", label="Cumulative ResNet18")
]

plt.legend(
    handles=legend_elements,
    loc="lower right"
)

# 제목
plt.title(
    "Cumulative CNN vs Cumulative ResNet18",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel("Test Accuracy", fontsize=15)

plt.ylim(0, 1.1)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()
```
